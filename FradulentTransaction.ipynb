```python
# ==========================================
# DBSCAN - FEATURE SELECTION
# ==========================================

features = [
    'step',
    'amount',
    'oldbalanceOrg',
    'newbalanceOrig',
    'oldbalanceDest',
    'newbalanceDest'
]

X = df[features].copy()

print("Selected numerical features:")
print(X.columns.tolist())
```

```python
# One-hot encode transaction type

type_encoded = pd.get_dummies(
    df['type'],
    prefix='type',
    dtype=int
)

X = pd.concat([X, type_encoded], axis=1)

print("\nFeatures after encoding:")
print(X.columns.tolist())

print("\nShape:")
print(X.shape)
```

```python
X = X.replace([np.inf, -np.inf], np.nan)

X = X.fillna(0)

print("Missing values:", X.isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Scaling completed.")
print("Scaled data shape:", X_scaled.shape)
```

```python
# ==========================================
# CREATE SAMPLE FOR DBSCAN
# ==========================================

sample_size = 100000

sample_indices = df.sample(
    n=sample_size,
    random_state=42
).index

X_sample = X_scaled[sample_indices]

df_sample = df.loc[sample_indices].copy()

print("X_sample shape:", X_sample.shape)
print("df_sample shape:", df_sample.shape)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.5,
    min_samples=10,
    n_jobs=-1
)

clusters = dbscan.fit_predict(X_sample)

df_sample['DBSCAN_Cluster'] = clusters
```

```python
print("Cluster distribution:")
print(
    df_sample['DBSCAN_Cluster']
    .value_counts()
    .sort_index()
)
```

```python
n_clusters = len(set(clusters)) - (1 if -1 in clusters else 0)

n_anomalies = (clusters == -1).sum()

print("\nNumber of clusters:", n_clusters)
print("Number of anomalies:", n_anomalies)
```

```python
anomalies = df_sample[
    df_sample['DBSCAN_Cluster'] == -1
]

print("Detected anomalies:", len(anomalies))

display(
    anomalies[
        [
            'step',
            'type',
            'amount',
            'oldbalanceOrg',
            'newbalanceOrig',
            'oldbalanceDest',
            'newbalanceDest',
            'isFraud'
        ]
    ].head(20)
)
```

```python
# DBSCAN anomaly prediction
df_sample['DBSCAN_Anomaly'] = (
    df_sample['DBSCAN_Cluster'] == -1
).astype(int)

# Compare DBSCAN anomaly with actual fraud
comparison = pd.crosstab(
    df_sample['DBSCAN_Anomaly'],
    df_sample['isFraud']
)

print(comparison)
```

```python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

y_actual = df_sample['isFraud']

y_predicted = df_sample['DBSCAN_Anomaly']

precision = precision_score(
    y_actual,
    y_predicted,
    zero_division=0
)

recall = recall_score(
    y_actual,
    y_predicted,
    zero_division=0
)

f1 = f1_score(
    y_actual,
    y_predicted,
    zero_division=0
)

print("Precision :", round(precision, 4))
print("Recall    :", round(recall, 4))
print("F1 Score  :", round(f1, 4))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(
    y_actual,
    y_predicted
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    xticklabels=['Normal', 'Anomaly'],
    yticklabels=['Normal', 'Fraud']
)

plt.xlabel("DBSCAN Prediction")
plt.ylabel("Actual Class")
plt.title("DBSCAN Anomaly Detection - Confusion Matrix")

plt.show()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_sample)

print("PCA completed.")
```

```python
plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=clusters,
    s=5,
    alpha=0.5
)

plt.title("DBSCAN Clustering of Financial Transactions")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.colorbar(
    scatter,
    label="Cluster"
)

plt.show()
```

```python
normal = clusters != -1
anomaly = clusters == -1

plt.figure(figsize=(10, 7))

plt.scatter(
    X_pca[normal, 0],
    X_pca[normal, 1],
    s=5,
    alpha=0.3,
    label='Clustered Transactions'
)

plt.scatter(
    X_pca[anomaly, 0],
    X_pca[anomaly, 1],
    s=15,
    alpha=0.8,
    label='DBSCAN Anomalies'
)

plt.title("DBSCAN-Based Fraud Anomaly Detection")

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.legend()

plt.show()
```
